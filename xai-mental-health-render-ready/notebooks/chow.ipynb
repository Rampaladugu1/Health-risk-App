```python
# ================================================================
# XAI-ENHANCED MENTAL HEALTH RISK ASSESSMENT PROTOTYPE
# Google Colab - Single Executable Cell
#
# Dissertation:
# "Explainable Artificial Intelligence for Mental Health Risk
#  Assessment in NHS Primary Care"
#
# Dataset:
# CDC/NHANES 2017-2018 public data
#
# Target:
# Elevated depression screening risk = PHQ-9 >= 10
#
# IMPORTANT:
# PHQ-9 item responses are NOT used as predictive features.
# They are used only to construct the target, preventing leakage.
# ================================================================

# ================================================================
# 1. INSTALL REQUIRED PACKAGES
# ================================================================

!pip -q install xgboost shap lime scikit-learn pandas numpy matplotlib seaborn joblib flask

# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import shap

from lime.lime_tabular import LimeTabularExplainer

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RandomizedSearchCV
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve,
    average_precision_score,
    brier_score_loss
)

from sklearn.calibration import calibration_curve

from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print("=" * 70)
print("XAI MENTAL HEALTH RISK ASSESSMENT")
print("=" * 70)
print("Libraries loaded successfully.")


# ================================================================
# 3. DOWNLOAD NHANES 2017-2018 DATA AUTOMATICALLY
# ================================================================

BASE_URL = "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2017/DataFiles/"

files = {
    "DEMO": "DEMO_J.xpt",   # Demographics
    "DPQ": "DPQ_J.xpt",     # Depression / PHQ-9
    "BMX": "BMX_J.xpt",     # Body measurements
    "SLQ": "SLQ_J.xpt",     # Sleep disorders
    "SMQ": "SMQ_J.xpt",     # Smoking
    "ALQ": "ALQ_J.xpt",     # Alcohol
    "PAQ": "PAQ_J.xpt",     # Physical activity
    "HUQ": "HUQ_J.xpt"      # Healthcare utilisation
}

data = {}

for name, filename in files.items():

    url = BASE_URL + filename

    try:
        data[name] = pd.read_sas(url, format="xport")

        print(
            f"{name:8s} loaded: "
            f"{data[name].shape[0]:,} rows x "
            f"{data[name].shape[1]:,} columns"
        )

    except Exception as e:

        print(f"WARNING: Could not load {name}: {e}")


# ================================================================
# 4. DISPLAY BASIC DATA INFORMATION
# ================================================================

print("\nAvailable datasets:")

for name, df in data.items():
    print(
        f"{name}: "
        f"{df.shape[0]:,} participants, "
        f"{df.shape[1]:,} variables"
    )


# ================================================================
# 5. CREATE PHQ-9 TARGET
# ================================================================

dpq = data["DPQ"].copy()

phq_items = [
    "DPQ010",
    "DPQ020",
    "DPQ030",
    "DPQ040",
    "DPQ050",
    "DPQ060",
    "DPQ070",
    "DPQ080",
    "DPQ090"
]

# Convert PHQ variables to numeric
for col in phq_items:
    dpq[col] = pd.to_numeric(dpq[col], errors="coerce")

# NHANES valid PHQ response values are 0-3.
# Values 7/9 are non-substantive responses.
for col in phq_items:
    dpq.loc[~dpq[col].isin([0, 1, 2, 3]), col] = np.nan

# Calculate PHQ-9 total score
dpq["PHQ9_TOTAL"] = dpq[phq_items].sum(axis=1, min_count=9)

# Clinical screening threshold:
# PHQ-9 >= 10 = elevated depression screening risk
dpq["DEPRESSION_RISK"] = (
    dpq["PHQ9_TOTAL"] >= 10
).astype(int)

print("\nPHQ-9 target created.")

print(
    dpq[
        ["PHQ9_TOTAL", "DEPRESSION_RISK"]
    ].describe()
)

print("\nTarget distribution:")
print(
    dpq["DEPRESSION_RISK"]
    .value_counts()
    .sort_index()
)

print("\nTarget percentages:")
print(
    dpq["DEPRESSION_RISK"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)


# ================================================================
# 6. MERGE DATASETS
# ================================================================

df = dpq[
    ["SEQN", "PHQ9_TOTAL", "DEPRESSION_RISK"]
].copy()

# Demographics
demo = data["DEMO"].copy()

demo_cols = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
    "RIDRETH1",
    "DMDEDUC2",
    "INDHHIN2",
    "DMDMARTL"
]

demo_cols = [
    c for c in demo_cols
    if c in demo.columns
]

df = df.merge(
    demo[demo_cols],
    on="SEQN",
    how="left"
)


# Body measurements
if "BMX" in data:

    bmx = data["BMX"].copy()

    bmx_cols = [
        "SEQN",
        "BMXBMI",
        "BMXWAIST"
    ]

    bmx_cols = [
        c for c in bmx_cols
        if c in bmx.columns
    ]

    df = df.merge(
        bmx[bmx_cols],
        on="SEQN",
        how="left"
    )


# Sleep
if "SLQ" in data:

    slq = data["SLQ"].copy()

    slq_cols = [
        "SEQN",
        "SLD012",
        "SLD013",
        "SLQ050"
    ]

    slq_cols = [
        c for c in slq_cols
        if c in slq.columns
    ]

    df = df.merge(
        slq[slq_cols],
        on="SEQN",
        how="left"
    )


# Smoking
if "SMQ" in data:

    smq = data["SMQ"].copy()

    smq_cols = [
        "SEQN",
        "SMQ020",
        "SMQ040"
    ]

    smq_cols = [
        c for c in smq_cols
        if c in smq.columns
    ]

    df = df.merge(
        smq[smq_cols],
        on="SEQN",
        how="left"
    )


# Alcohol
if "ALQ" in data:

    alq = data["ALQ"].copy()

    alq_cols = [
        "SEQN",
        "ALQ101",
        "ALQ110"
    ]

    alq_cols = [
        c for c in alq_cols
        if c in alq.columns
    ]

    df = df.merge(
        alq[alq_cols],
        on="SEQN",
        how="left"
    )


# Physical activity
if "PAQ" in data:

    paq = data["PAQ"].copy()

    paq_cols = [
        "SEQN",
        "PAQ605",
        "PAQ620",
        "PAD615",
        "PAD630"
    ]

    paq_cols = [
        c for c in paq_cols
        if c in paq.columns
    ]

    df = df.merge(
        paq[paq_cols],
        on="SEQN",
        how="left"
    )


print("\nMerged dataset:")
print(
    f"Rows: {df.shape[0]:,}"
)

print(
    f"Columns: {df.shape[1]:,}"
)

print("\nColumns:")
print(df.columns.tolist())


# ================================================================
# 7. CLEAN DATA
# ================================================================

# Replace NHANES missing/non-response codes
MISSING_CODES = [
    7,
    9,
    77,
    99,
    777,
    999
]

for col in df.columns:

    if col not in [
        "SEQN",
        "PHQ9_TOTAL",
        "DEPRESSION_RISK"
    ]:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        df[col] = df[col].replace(
            MISSING_CODES,
            np.nan
        )


# Keep adults
df = df[
    df["RIDAGEYR"] >= 18
].copy()


# Remove rows without valid target
df = df.dropna(
    subset=["DEPRESSION_RISK"]
)


print("\nClean dataset:")
print(
    f"{df.shape[0]:,} adult participants"
)


# ================================================================
# 8. EXPLORATORY DATA ANALYSIS
# ================================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="DEPRESSION_RISK"
)

plt.title(
    "Distribution of Elevated Depression Screening Risk"
)

plt.xlabel(
    "Depression risk (PHQ-9 ≥ 10)"
)

plt.ylabel(
    "Number of participants"
)

plt.tight_layout()

plt.show()


# PHQ-9 distribution

plt.figure(figsize=(9, 5))

sns.histplot(
    df["PHQ9_TOTAL"].dropna(),
    bins=28,
    kde=True
)

plt.axvline(
    10,
    linestyle="--",
    linewidth=2
)

plt.title(
    "PHQ-9 Score Distribution"
)

plt.xlabel(
    "PHQ-9 total score"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ================================================================
# 9. DEFINE MODEL FEATURES
# ================================================================

# IMPORTANT:
# PHQ9_TOTAL and the nine PHQ questions are NOT features.
# This prevents target leakage.

candidate_features = [
    "RIAGENDR",
    "RIDAGEYR",
    "RIDRETH1",
    "DMDEDUC2",
    "INDHHIN2",
    "DMDMARTL",
    "BMXBMI",
    "BMXWAIST",
    "SLD012",
    "SLD013",
    "SLQ050",
    "SMQ020",
    "SMQ040",
    "ALQ101",
    "ALQ110",
    "PAQ605",
    "PAQ620",
    "PAD615",
    "PAD630"
]

features = [
    c for c in candidate_features
    if c in df.columns
]

X = df[features].copy()

y = df["DEPRESSION_RISK"].astype(int)


print("\nFeatures used:")
for feature in features:
    print(" -", feature)

print(
    "\nNumber of features:",
    len(features)
)


# ================================================================
# 10. IDENTIFY NUMERIC AND CATEGORICAL VARIABLES
# ================================================================

categorical_features = [
    "RIAGENDR",
    "RIDRETH1",
    "DMDEDUC2",
    "INDHHIN2",
    "DMDMARTL",
    "SLQ050",
    "SMQ020",
    "SMQ040",
    "PAQ605",
    "PAQ620"
]

categorical_features = [
    c for c in categorical_features
    if c in X.columns
]

numeric_features = [
    c for c in X.columns
    if c not in categorical_features
]


# ================================================================
# 11. PREPROCESSING PIPELINE
# ================================================================

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],
    remainder="drop"
)


# ================================================================
# 12. TRAIN / TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=RANDOM_STATE
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ================================================================
# 13. TRANSFORM DATA
# ================================================================

X_train_processed = preprocessor.fit_transform(
    X_train
)

X_test_processed = preprocessor.transform(
    X_test
)

feature_names = (
    preprocessor
    .get_feature_names_out()
)

feature_names = [
    name.replace(
        "numeric__",
        ""
    ).replace(
        "categorical__",
        ""
    )
    for name in feature_names
]

X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)


print(
    "\nProcessed feature matrix:",
    X_train_processed.shape
)


# ================================================================
# 14. HANDLE CLASS IMBALANCE
# ================================================================

negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print(
    "\nClass imbalance ratio:",
    round(scale_pos_weight, 3)
)


# ================================================================
# 15. BASELINE MODEL - LOGISTIC REGRESSION
# ================================================================

logistic_model = LogisticRegression(
    max_iter=2000,
    class_weight="balanced",
    random_state=RANDOM_STATE
)

logistic_model.fit(
    X_train_processed,
    y_train
)

logistic_prob = logistic_model.predict_proba(
    X_test_processed
)[:, 1]

logistic_pred = (
    logistic_prob >= 0.5
).astype(int)


# ================================================================
# 16. RANDOM FOREST
# ================================================================

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=5,
    class_weight="balanced",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

rf_model.fit(
    X_train_processed,
    y_train
)

rf_prob = rf_model.predict_proba(
    X_test_processed
)[:, 1]

rf_pred = (
    rf_prob >= 0.5
).astype(int)


# ================================================================
# 17. XGBOOST MODEL
# ================================================================

xgb_model = XGBClassifier(
    n_estimators=400,
    max_depth=4,
    learning_rate=0.04,
    subsample=0.85,
    colsample_bytree=0.85,
    min_child_weight=3,
    reg_lambda=2,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=scale_pos_weight,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

xgb_model.fit(
    X_train_processed,
    y_train
)

xgb_prob = xgb_model.predict_proba(
    X_test_processed
)[:, 1]

xgb_pred = (
    xgb_prob >= 0.5
).astype(int)


# ================================================================
# 18. MODEL EVALUATION FUNCTION
# ================================================================

def evaluate_model(
    name,
    y_true,
    prediction,
    probability
):

    return {
        "Model": name,
        "Accuracy": accuracy_score(
            y_true,
            prediction
        ),
        "Precision": precision_score(
            y_true,
            prediction,
            zero_division=0
        ),
        "Recall / Sensitivity": recall_score(
            y_true,
            prediction,
            zero_division=0
        ),
        "F1": f1_score(
            y_true,
            prediction,
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            y_true,
            probability
        ),
        "PR-AUC": average_precision_score(
            y_true,
            probability
        ),
        "Brier Score": brier_score_loss(
            y_true,
            probability
        )
    }


results = pd.DataFrame([
    evaluate_model(
        "Logistic Regression",
        y_test,
        logistic_pred,
        logistic_prob
    ),

    evaluate_model(
        "Random Forest",
        y_test,
        rf_pred,
        rf_prob
    ),

    evaluate_model(
        "XGBoost",
        y_test,
        xgb_pred,
        xgb_prob
    )
])


print("\n")
print("=" * 70)
print("MODEL PERFORMANCE")
print("=" * 70)

display(
    results.round(4)
)


# ================================================================
# 19. CLASSIFICATION REPORT FOR XGBOOST
# ================================================================

print("\nXGBoost Classification Report:")
print(
    classification_report(
        y_test,
        xgb_pred,
        target_names=[
            "Lower risk",
            "Elevated risk"
        ],
        zero_division=0
    )
)


# ================================================================
# 20. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    xgb_pred
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Lower risk",
        "Elevated risk"
    ],
    yticklabels=[
        "Lower risk",
        "Elevated risk"
    ]
)

plt.title(
    "XGBoost Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.tight_layout()

plt.show()


# ================================================================
# 21. ROC CURVES
# ================================================================

plt.figure(figsize=(8, 6))

for name, probability in [
    ("Logistic Regression", logistic_prob),
    ("Random Forest", rf_prob),
    ("XGBoost", xgb_prob)
]:

    fpr, tpr, _ = roc_curve(
        y_test,
        probability
    )

    auc = roc_auc_score(
        y_test,
        probability
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{name} (AUC={auc:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve Comparison"
)

plt.legend()

plt.tight_layout()

plt.show()


# ================================================================
# 22. PRECISION-RECALL CURVE
# ================================================================

precision, recall, thresholds = precision_recall_curve(
    y_test,
    xgb_prob
)

pr_auc = average_precision_score(
    y_test,
    xgb_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall,
    precision,
    linewidth=2,
    label=f"XGBoost PR-AUC={pr_auc:.3f}"
)

plt.xlabel(
    "Recall / Sensitivity"
)

plt.ylabel(
    "Precision / PPV"
)

plt.title(
    "XGBoost Precision-Recall Curve"
)

plt.legend()

plt.tight_layout()

plt.show()


# ================================================================
# 23. CALIBRATION ANALYSIS
# ================================================================

fraction_pos, mean_predicted = calibration_curve(
    y_test,
    xgb_prob,
    n_bins=10,
    strategy="uniform"
)

plt.figure(figsize=(7, 6))

plt.plot(
    mean_predicted,
    fraction_pos,
    marker="o",
    linewidth=2,
    label="XGBoost"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect calibration"
)

plt.xlabel(
    "Mean predicted probability"
)

plt.ylabel(
    "Observed proportion"
)

plt.title(
    "XGBoost Calibration Curve"
)

plt.legend()

plt.tight_layout()

plt.show()

print(
    "XGBoost Brier score:",
    round(
        brier_score_loss(
            y_test,
            xgb_prob
        ),
        4
    )
)


# ================================================================
# 24. SHAP EXPLAINABILITY
# ================================================================

print("\n")
print("=" * 70)
print("SHAP EXPLAINABILITY")
print("=" * 70)

explainer = shap.TreeExplainer(
    xgb_model
)

shap_values = explainer.shap_values(
    X_test_processed
)

# SHAP summary plot
plt.figure()

shap.summary_plot(
    shap_values,
    X_test_processed,
    show=False
)

plt.title(
    "SHAP Feature Importance - XGBoost"
)

plt.tight_layout()

plt.show()


# ================================================================
# 25. SHAP BAR PLOT
# ================================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_test_processed,
    plot_type="bar",
    show=False
)

plt.title(
    "Mean Absolute SHAP Feature Importance"
)

plt.tight_layout()

plt.show()


# ================================================================
# 26. SHAP WATERFALL EXPLANATION FOR ONE PATIENT
# ================================================================

patient_position = 0

patient_features = X_test_processed.iloc[
    patient_position
]

patient_original = X_test.iloc[
    patient_position
]

patient_probability = xgb_prob[
    patient_position
]

print("\nExample patient:")
display(
    patient_original.to_frame(
        "Value"
    )
)

print(
    "\nPredicted elevated-risk probability:",
    round(
        patient_probability,
        4
    )
)

if patient_probability >= 0.5:
    print("Risk classification: ELEVATED RISK")
else:
    print("Risk classification: LOWER RISK")


# SHAP waterfall
shap_explanation = shap.Explanation(
    values=shap_values[
        patient_position
    ],
    base_values=explainer.expected_value,
    data=patient_features.values,
    feature_names=patient_features.index.tolist()
)

shap.plots.waterfall(
    shap_explanation,
    max_display=12
)


# ================================================================
# 27. TOP SHAP FACTORS FOR PATIENT
# ================================================================

patient_shap = pd.DataFrame({
    "Feature": X_test_processed.columns,
    "SHAP_Value": shap_values[
        patient_position
    ],
    "Absolute_SHAP": np.abs(
        shap_values[
            patient_position
        ]
    )
})

patient_shap = patient_shap.sort_values(
    "Absolute_SHAP",
    ascending=False
)

print("\nTop factors influencing this prediction:")

display(
    patient_shap[
        [
            "Feature",
            "SHAP_Value"
        ]
    ].head(10).round(4)
)


# ================================================================
# 28. LIME EXPLANATION
# ================================================================

print("\n")
print("=" * 70)
print("LIME EXPLANATION")
print("=" * 70)

lime_explainer = LimeTabularExplainer(
    X_train_processed.values,
    feature_names=X_train_processed.columns.tolist(),
    class_names=[
        "Lower risk",
        "Elevated risk"
    ],
    mode="classification",
    discretize_continuous=True,
    random_state=RANDOM_STATE
)


def lime_predict(x):

    return xgb_model.predict_proba(
        x
    )


lime_explanation = lime_explainer.explain_instance(
    X_test_processed.iloc[
        patient_position
    ].values,
    lime_predict,
    num_features=10
)

lime_list = lime_explanation.as_list()

lime_df = pd.DataFrame(
    lime_list,
    columns=[
        "Feature",
        "LIME_Weight"
    ]
)

display(
    lime_df
)

print(
    "\nLIME explanation:"
)

for feature, weight in lime_list:
    direction = (
        "increases"
        if weight > 0
        else "decreases"
    )

    print(
        f"{feature}: {direction} "
        f"elevated-risk prediction "
        f"(weight={weight:.4f})"
    )


# ================================================================
# 29. SHAP VS LIME STABILITY ANALYSIS
# ================================================================

print("\n")
print("=" * 70)
print("SHAP VS LIME STABILITY")
print("=" * 70)

# Select several test patients
n_patients = min(
    30,
    len(X_test_processed)
)

shap_rankings = []
lime_rankings = []

for i in range(n_patients):

    # SHAP ranking
    shap_order = (
        pd.Series(
            np.abs(
                shap_values[i]
            ),
            index=X_test_processed.columns
        )
        .sort_values(
            ascending=False
        )
        .head(10)
        .index
        .tolist()
    )

    shap_rankings.append(
        shap_order
    )

    # LIME ranking
    exp = lime_explainer.explain_instance(
        X_test_processed.iloc[i].values,
        lime_predict,
        num_features=10
    )

    lime_order = [
        item[0]
        for item in exp.as_list()
    ]

    lime_rankings.append(
        lime_order
    )


# Calculate average overlap of top features
overlaps = []

for s, l in zip(
    shap_rankings,
    lime_rankings
):

    overlap = len(
        set(s).intersection(
            set(l)
        )
    ) / 10

    overlaps.append(
        overlap
    )

mean_overlap = np.mean(
    overlaps
)

print(
    "Mean top-10 feature overlap "
    "between SHAP and LIME:",
    round(
        mean_overlap,
        4
    )
)

print(
    "\nInterpretation:"
)

print(
    "Higher overlap indicates greater agreement "
    "between the two explanation methods."
)


# ================================================================
# 30. GLOBAL FEATURE IMPORTANCE
# ================================================================

global_shap = pd.DataFrame({
    "Feature": X_test_processed.columns,
    "Mean_Absolute_SHAP": np.abs(
        shap_values
    ).mean(axis=0)
})

global_shap = global_shap.sort_values(
    "Mean_Absolute_SHAP",
    ascending=False
)

print("\nTop global SHAP features:")

display(
    global_shap.head(15).round(4)
)


# ================================================================
# 31. FAIRNESS ANALYSIS
# ================================================================

print("\n")
print("=" * 70)
print("FAIRNESS ANALYSIS")
print("=" * 70)

# Add predictions back to test data
fairness_df = X_test.copy()

fairness_df["Actual"] = y_test.values
fairness_df["Probability"] = xgb_prob
fairness_df["Prediction"] = xgb_pred


def subgroup_metrics(
    data,
    group_column
):

    rows = []

    if group_column not in data.columns:
        return pd.DataFrame()

    for group in sorted(
        data[group_column].dropna().unique()
    ):

        subset = data[
            data[group_column] == group
        ]

        if len(subset) < 10:
            continue

        actual = subset["Actual"]
        prediction = subset["Prediction"]
        probability = subset["Probability"]

        tn, fp, fn, tp = confusion_matrix(
            actual,
            prediction,
            labels=[0, 1]
        ).ravel()

        sensitivity = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else np.nan
        )

        specificity = (
            tn / (tn + fp)
            if (tn + fp) > 0
            else np.nan
        )

        ppv = (
            tp / (tp + fp)
            if (tp + fp) > 0
            else np.nan
        )

        npv = (
            tn / (tn + fn)
            if (tn + fn) > 0
            else np.nan
        )

        rows.append({
            "Group": group,
            "N": len(subset),
            "Prevalence": actual.mean(),
            "Sensitivity": sensitivity,
            "Specificity": specificity,
            "PPV": ppv,
            "NPV": npv,
            "Mean_Predicted_Risk": probability.mean()
        })

    return pd.DataFrame(rows)


# Sex fairness
sex_fairness = subgroup_metrics(
    fairness_df,
    "RIAGENDR"
)

print("\nFairness by sex:")
display(
    sex_fairness.round(4)
)


# Race/ethnicity fairness
race_fairness = subgroup_metrics(
    fairness_df,
    "RIDRETH1"
)

print("\nFairness by race/ethnicity:")
display(
    race_fairness.round(4)
)


# ================================================================
# 32. AGE-GROUP FAIRNESS
# ================================================================

fairness_df["AGE_GROUP"] = pd.cut(
    fairness_df["RIDAGEYR"],
    bins=[
        17,
        29,
        44,
        59,
        150
    ],
    labels=[
        "18-29",
        "30-44",
        "45-59",
        "60+"
    ]
)

age_fairness = subgroup_metrics(
    fairness_df,
    "AGE_GROUP"
)

print("\nFairness by age group:")
display(
    age_fairness.round(4)
)


# ================================================================
# 33. FAIRNESS VISUALISATION
# ================================================================

if not sex_fairness.empty:

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=sex_fairness,
        x="Group",
        y="Sensitivity"
    )

    plt.title(
        "XGBoost Sensitivity by Sex Group"
    )

    plt.xlabel(
        "NHANES gender code"
    )

    plt.ylabel(
        "Sensitivity"
    )

    plt.tight_layout()

    plt.show()


# ================================================================
# 34. ERROR ANALYSIS
# ================================================================

error_df = X_test.copy()

error_df["Actual"] = y_test.values
error_df["Prediction"] = xgb_pred
error_df["Probability"] = xgb_prob

error_df["Error_Type"] = "Correct"

error_df.loc[
    (error_df["Actual"] == 0) &
    (error_df["Prediction"] == 1),
    "Error_Type"
] = "False Positive"

error_df.loc[
    (error_df["Actual"] == 1) &
    (error_df["Prediction"] == 0),
    "Error_Type"
] = "False Negative"

print("\nError distribution:")
print(
    error_df["Error_Type"]
    .value_counts()
)


# ================================================================
# 35. FIND HIGH-RISK EXAMPLE PATIENTS
# ================================================================

example_predictions = X_test.copy()

example_predictions[
    "Predicted_Risk"
] = xgb_prob

example_predictions[
    "Classification"
] = np.where(
    xgb_prob >= 0.5,
    "Elevated Risk",
    "Lower Risk"
)

example_predictions = (
    example_predictions
    .sort_values(
        "Predicted_Risk",
        ascending=False
    )
)

print("\nHighest predicted-risk examples:")

display(
    example_predictions.head(5)
)


# ================================================================
# 36. CLINICIAN-FACING RISK FUNCTION
# ================================================================

def assess_patient(
    patient_data,
    threshold=0.50
):

    """
    Prototype clinical decision-support function.

    This is NOT a diagnostic tool.
    It demonstrates how a clinician-facing
    XAI prediction could be generated.
    """

    patient_df = pd.DataFrame(
        [patient_data]
    )

    patient_processed = (
        preprocessor.transform(
            patient_df
        )
    )

    patient_processed = pd.DataFrame(
        patient_processed,
        columns=feature_names
    )

    probability = (
        xgb_model.predict_proba(
            patient_processed
        )[0, 1]
    )

    prediction = int(
        probability >= threshold
    )

    patient_shap_values = (
        explainer.shap_values(
            patient_processed
        )[0]
    )

    explanation = pd.DataFrame({
        "Feature": feature_names,
        "SHAP_Value": patient_shap_values
    })

    explanation[        "Absolute_SHAP"
    ] = np.abs(
        explanation[
            "SHAP_Value"
        ]
    )

    explanation = (
        explanation
        .sort_values(
            "Absolute_SHAP",
            ascending=False
        )
        .head(10)
    )

    return {
        "risk_probability": probability,
        "classification": (
            "Elevated depression screening risk"
            if prediction == 1
            else
            "Lower depression screening risk"
        ),
        "top_explanatory_features": explanation
    }


# ================================================================
# 37. TEST THE CLINICIAN-FACING FUNCTION
# ================================================================

example_patient = {}

for col in features:

    value = X_test.iloc[
        0
    ][col]

    if pd.isna(value):

        if col in numeric_features:

            value = X_train[col].median()

        else:

            value = X_train[col].mode()[0]

    example_patient[col] = value


assessment = assess_patient(
    example_patient
)

print("\n")
print("=" * 70)
print("CLINICIAN-FACING PROTOTYPE")
print("=" * 70)

print(
    "\nPredicted risk probability:",
    f"{assessment['risk_probability']:.2%}"
)

print(
    "Classification:",
    assessment["classification"]
)

print(
    "\nTop explanatory factors:"
)

display(
    assessment[
        "top_explanatory_features"
    ][
        [
            "Feature",
            "SHAP_Value"
        ]
    ].round(4)
)


# ================================================================
# 38. SAVE MODEL AND XAI COMPONENTS
# ================================================================

os.makedirs(
    "/content/xai_mental_health_model",
    exist_ok=True
)

joblib.dump(
    xgb_model,
    "/content/xai_mental_health_model/xgboost_model.joblib"
)

joblib.dump(
    preprocessor,
    "/content/xai_mental_health_model/preprocessor.joblib"
)

joblib.dump(
    explainer,
    "/content/xai_mental_health_model/shap_explainer.joblib"
)

joblib.dump(
    feature_names,
    "/content/xai_mental_health_model/feature_names.joblib"
)

joblib.dump(
    X_train,
    "/content/xai_mental_health_model/X_train.joblib"
) # Save X_train

joblib.dump(
    features,
    "/content/xai_mental_health_model/features.joblib"
) # Save features list

joblib.dump(
    numeric_features,
    "/content/xai_mental_health_model/numeric_features.joblib"
) # Save numeric_features list

joblib.dump(
    categorical_features,
    "/content/xai_mental_health_model/categorical_features.joblib"
) # Save categorical_features list

results.to_csv(
    "/content/xai_mental_health_model/model_results.csv",
    index=False
)

global_shap.to_csv(
    "/content/xai_mental_health_model/shap_feature_importance.csv",
    index=False
)

sex_fairness.to_csv(
    "/content/xai_mental_health_model/fairness_sex.csv",
    index=False
)

race_fairness.to_csv(
    "/content/xai_mental_health_model/fairness_race.csv",
    index=False
)

age_fairness.to_csv(
    "/content/xai_mental_health_model/fairness_age.csv",
    index=False
)


# ================================================================
# 39. SAVE TEST DATA
# ================================================================

results_test = X_test.copy()

results_test[
    "Actual"
] = y_test.values

results_test[
    "Predicted_Probability"
] = xgb_prob

results_test[
    "Predicted_Class"
] = xgb_pred

results_test.to_csv(
    "/content/xai_mental_health_model/test_predictions.csv",
    index=False
)


# ================================================================
# 40. FINAL SUMMARY
# ================================================================

best_model_row = results.loc[
    results["ROC-AUC"].idxmax()
]

print("\n")
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    f"""
Best model according to ROC-AUC:
{best_model_row['Model']}

ROC-AUC:
{best_model_row['ROC-AUC']:.4f}

F1:
{best_model_row['F1']:.4f}

Sensitivity:
{best_model_row['Recall / Sensitivity']:.4f}

Precision:
{best_model_row['Precision']:.4f}

Accuracy:
{best_model_row['Accuracy']:.4f}

PR-AUC:
{best_model_row['PR-AUC']:.4f}

Brier Score:
{best_model_row['Brier Score']:.4f}

SHAP/LIME top-10 feature overlap:
{mean_overlap:.4f}

Number of model features:
{len(feature_names)}

Training participants:
{len(X_train):,}

Testing participants:
{len(X_test):,}
"""
)

print("=" * 70)

print(
    "\nFiles saved to:"
)

print(
    "/content/xai_mental_health_model/"
)

print(
    "\nIMPORTANT:"
)

print(
    "This is a research prototype and not a clinical diagnostic system."
)

print(
    "The dataset is US NHANES data, so results should NOT be interpreted"
)

print(
    "as evidence of performance in NHS primary care without external validation."
)

print(
    "\nDONE."
)
```

```python
# ================================================================
# XAI MENTAL HEALTH RISK ASSESSMENT
# COMPLETE END-TO-END GOOGLE COLAB PIPELINE
# ================================================================
#
# Dissertation:
# "Explainable Artificial Intelligence for Mental Health Risk
#  Assessment in NHS Primary Care"
#
# Target:
# Elevated depressive symptom risk = PHQ-9 total >= 10
#
# Dataset:
# CDC/NHANES 2017-2018
#
# Models:
# 1. Logistic Regression
# 2. Random Forest
# 3. XGBoost
#
# Explainability:
# 1. SHAP
# 2. LIME
# 3. SHAP/LIME agreement
# 4. LIME stability
#
# Evaluation:
# ROC-AUC
# PR-AUC
# Accuracy
# Precision / PPV
# Sensitivity / Recall
# Specificity
# F1
# NPV
# Brier Score
# Calibration
# Fairness
#
# IMPORTANT:
# This is a research prototype.
# It is NOT a clinical diagnostic system.
# ================================================================


# ================================================================
# 1. INSTALL REQUIRED PACKAGES
# ================================================================

!pip -q install \
    xgboost \
    shap \
    lime \
    scikit-learn \
    pandas \
    numpy \
    matplotlib \
    seaborn \
    joblib


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import os
import re
import json
import warnings
import urllib.request

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import joblib
import shap

from io import BytesIO

from lime.lime_tabular import LimeTabularExplainer

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    confusion_matrix,
    roc_curve,
    precision_recall_curve
)

from sklearn.calibration import calibration_curve

import xgboost as xgb


warnings.filterwarnings("ignore")


# ================================================================
# 3. GLOBAL SETTINGS
# ================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)


OUTPUT_DIR = "/content/xai_mental_health_model"


os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


print("=" * 70)
print("XAI MENTAL HEALTH RISK ASSESSMENT")
print("COMPLETE PIPELINE")
print("=" * 70)

print(
    f"\nOutput directory: {OUTPUT_DIR}"
)


# ================================================================
# 4. DOWNLOAD FUNCTION
# ================================================================

BASE_URL = (
    "https://wwwn.cdc.gov/"
    "Nchs/Data/Nhanes/Public/"
    "2017/DataFiles/"
)


FILES = {

    "DEMO":
        "DEMO_J.xpt",

    "DPQ":
        "DPQ_J.xpt",

    "BMX":
        "BMX_J.xpt",

    "SLQ":
        "SLQ_J.xpt",

    "SMQ":
        "SMQ_J.xpt",

    "ALQ":
        "ALQ_J.xpt",

    "PAQ":
        "PAQ_J.xpt",

    "HUQ":
        "HUQ_J.xpt"

}


def download_xpt(
    filename
):

    url = BASE_URL + filename

    print(
        f"Downloading: {filename}"
    )

    response = urllib.request.urlopen(
        url
    )

    data = response.read()

    df = pd.read_sas(
        BytesIO(data),
        format="xport"
    )

    print(
        f"  ✓ {df.shape[0]:,} rows "
        f"x {df.shape[1]:,} columns"
    )

    return df


# ================================================================
# 5. DOWNLOAD NHANES DATA
# ================================================================

print("\n")
print("=" * 70)
print("DOWNLOADING NHANES 2017-2018 DATA")
print("=" * 70)


datasets = {}


for key, filename in FILES.items():

    datasets[key] = download_xpt(
        filename
    )


demo = datasets["DEMO"].copy()

dpq = datasets["DPQ"].copy()

bmx = datasets["BMX"].copy()

slq = datasets["SLQ"].copy()

smq = datasets["SMQ"].copy()

alq = datasets["ALQ"].copy()

paq = datasets["PAQ"].copy()

huq = datasets["HUQ"].copy()


# ================================================================
# 6. BASIC DATA VALIDATION
# ================================================================

print("\n")
print("=" * 70)
print("BASIC DATA VALIDATION")
print("=" * 70)


for name, df in datasets.items():

    print(
        f"{name:6s}: "
        f"{len(df):,} participants, "
        f"{len(df.columns):,} variables"
    )


if "SEQN" not in dpq.columns:

    raise RuntimeError(
        "SEQN is missing from DPQ dataset."
    )


# ================================================================
# 7. PHQ-9 VARIABLES
# ================================================================

PHQ_ITEMS = [

    "DPQ010",
    "DPQ020",
    "DPQ030",
    "DPQ040",
    "DPQ050",
    "DPQ060",
    "DPQ070",
    "DPQ080",
    "DPQ090"

]


print("\n")
print("=" * 70)
print("PHQ-9 VALIDATION")
print("=" * 70)


missing_phq_items = [

    col
    for col in PHQ_ITEMS
    if col not in dpq.columns

]


if missing_phq_items:

    raise RuntimeError(
        "Missing PHQ-9 variables: "
        + str(missing_phq_items)
    )


print(
    "✓ All nine PHQ-9 items found."
)


# ================================================================
# 8. ROBUST NHANES VALUE CONVERSION
# ================================================================
#
# IMPORTANT:
#
# NHANES SAS XPT files may contain extremely small floating-point
# representations of zero:
#
# 5.397605e-79
#
# This represents SAS missing/zero encoding in the imported file
# and MUST be converted to 0.
#
# Valid PHQ-9 responses are:
#
# 0 = Not at all
# 1 = Several days
# 2 = More than half the days
# 3 = Nearly every day
#
# 7 = Refused
# 9 = Don't know
#
# Missing/refused/don't know are excluded.
# ================================================================


def clean_nhanes_phq(
    series
):

    values = []


    for value in series:

        if pd.isna(value):

            values.append(
                np.nan
            )

            continue


        # Bytes

        if isinstance(
            value,
            bytes
        ):

            try:

                value = value.decode(
                    "utf-8",
                    errors="ignore"
                )

            except:

                value = str(
                    value
                )


        # Strings

        if isinstance(
            value,
            str
        ):

            value = value.strip()

            value = (
                value
                .replace(
                    "b'",
                    ""
                )
                .replace(
                    'b"',
                    ""
                )
                .strip(
                    "'\""
                )
            )


        try:

            numeric_value = float(
                value
            )

        except:

            values.append(
                np.nan
            )

            continue


        # SAS imported representation of zero

        if abs(
            numeric_value
        ) < 1e-20:

            numeric_value = 0.0


        # Valid PHQ response

        if numeric_value in [
            0,
            1,
            2,
            3
        ]:

            values.append(
                float(
                    numeric_value
                )
            )

        else:

            values.append(
                np.nan
            )


    return pd.Series(
        values,
        index=series.index,
        dtype="float64"
    )


# ================================================================
# 9. CLEAN ALL PHQ-9 ITEMS
# ================================================================

print("\nCleaning PHQ-9 responses...")


for item in PHQ_ITEMS:

    dpq[
        item + "_CLEAN"
    ] = clean_nhanes_phq(
        dpq[item]
    )


# ================================================================
# 10. DISPLAY CLEANED DISTRIBUTIONS
# ================================================================

print("\n")
print("=" * 70)
print("CLEANED PHQ-9 VALUE DISTRIBUTIONS")
print("=" * 70)


for item in PHQ_ITEMS:

    clean_col = item + "_CLEAN"

    print(
        f"\n{item}"
    )

    print(
        dpq[
            clean_col
        ].value_counts(
            dropna=False
        ).sort_index()
    )


# ================================================================
# 11. VALID RESPONSE COUNTS
# ================================================================

print("\n")
print("=" * 70)
print("VALID PHQ-9 RESPONSE COUNTS")
print("=" * 70)


valid_counts = {}


for item in PHQ_ITEMS:

    clean_col = item + "_CLEAN"

    count = int(
        dpq[
            clean_col
        ].notna().sum()
    )


    valid_counts[
        item
    ] = count


    percentage = (

        count
        /
        len(dpq)
        *
        100

    )


    print(
        f"{item}: "
        f"{count:,} valid "
        f"({percentage:.1f}%)"
    )


# ================================================================
# 12. CALCULATE PHQ-9 TOTAL
# ================================================================

clean_columns = [

    item + "_CLEAN"
    for item in PHQ_ITEMS

]


# Require all nine items

complete_phq_mask = (

    dpq[
        clean_columns
    ]
    .notna()
    .all(axis=1)

)


dpq[
    "PHQ9_TOTAL"
] = np.nan


dpq.loc[
    complete_phq_mask,
    "PHQ9_TOTAL"
] = (

    dpq.loc[
        complete_phq_mask,
        clean_columns
    ]
    .sum(
        axis=1
    )

)


# ================================================================
# 13. PHQ-9 COMPLETE CASE VALIDATION
# ================================================================

complete_phq = int(
    complete_phq_mask.sum()
)


print("\n")
print("=" * 70)
print("PHQ-9 COMPLETE CASE VALIDATION")
print("=" * 70)


print(
    f"Total DPQ participants: "
    f"{len(dpq):,}"
)


print(
    f"Complete PHQ-9 cases: "
    f"{complete_phq:,}"
)


print(
    f"Complete PHQ-9 percentage: "
    f"{complete_phq / len(dpq) * 100:.2f}%"
)


if complete_phq == 0:

    raise RuntimeError(
        "No complete PHQ-9 cases were found."
    )


phq_min = dpq[
    "PHQ9_TOTAL"
].min()


phq_max = dpq[
    "PHQ9_TOTAL"
].max()


print(
    f"PHQ-9 minimum: "
    f"{phq_min}"
)


print(
    f"PHQ-9 maximum: "
    f"{phq_max}"
)


# ================================================================
# 14. CRITICAL PHQ VALIDATION
# ================================================================

if complete_phq < 1000:

    raise RuntimeError(
        "\n\n"
        "PHQ-9 VALIDATION FAILED.\n"
        "Fewer than 1,000 complete PHQ-9 cases were found.\n"
        "The program stopped before machine learning."
    )


if phq_min < 0 or phq_max > 27:

    raise RuntimeError(
        "PHQ-9 total outside valid range 0-27."
    )


print(
    "\n✓ PHQ-9 validation PASSED."
)


# ================================================================
# 15. CREATE DEPRESSION RISK TARGET
# ================================================================
#
# PHQ-9 >= 10 is commonly used as a threshold for clinically
# significant depressive symptoms / elevated depression risk.
#
# This is NOT the same as a confirmed clinical diagnosis.
# ================================================================


dpq[
    "DEPRESSION_RISK"
] = np.nan


dpq.loc[
    complete_phq_mask,
    "DEPRESSION_RISK"
] = (

    dpq.loc[
        complete_phq_mask,
        "PHQ9_TOTAL"
    ]
    >= 10

).astype(int)


negative_cases = int(

    (
        dpq[
            "DEPRESSION_RISK"
        ]
        == 0
    )
    .sum()

)


positive_cases = int(

    (
        dpq[
            "DEPRESSION_RISK"
        ]
        == 1
    )
    .sum()

)


prevalence = (

    positive_cases
    /
    complete_phq
    *
    100

)


print("\n")
print("=" * 70)
print("TARGET VALIDATION")
print("=" * 70)


print(
    f"Negative cases: "
    f"{negative_cases:,}"
)


print(
    f"Positive cases: "
    f"{positive_cases:,}"
)


print(
    f"Positive prevalence: "
    f"{prevalence:.2f}%"
)


if positive_cases < 100:

    raise RuntimeError(
        "Too few positive cases for reliable modelling."
    )


# ================================================================
# 16. MERGE DATASETS
# ================================================================

print("\n")
print("=" * 70)
print("MERGING NHANES DATASETS")
print("=" * 70)


# Keep only variables required for modelling.
#
# This avoids creating a huge unnecessary dataframe.

demo_keep = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
    "RIDRETH1",
    "DMDEDUC2",
    "INDHHIN2",
    "DMDMARTL"
]


bmx_keep = [
    "SEQN",
    "BMXBMI",
    "BMXWAIST"
]


slq_keep = [
    "SEQN",
    "SLD012",
    "SLD013",
    "SLQ050"
]


smq_keep = [
    "SEQN",
    "SMQ020",
    "SMQ040"
]


paq_keep = [
    "SEQN",
    "PAQ605",
    "PAQ620",
    "PAD615",
    "PAD630"
]


huq_keep = [
    "SEQN",
    "HUQ010",
    "HUQ030"
]


def keep_available(
    df,
    columns
):

    return df[
        [
            col
            for col in columns
            if col in df.columns
        ]
    ].copy()


demo_small = keep_available(
    demo,
    demo_keep
)


bmx_small = keep_available(
    bmx,
    bmx_keep
)


slq_small = keep_available(
    slq,
    slq_keep
)


smq_small = keep_available(
    smq,
    smq_keep
)


paq_small = keep_available(
    paq,
    paq_keep
)


huq_small = keep_available(
    huq,
    huq_keep
)


# Start with PHQ dataset

model_df = dpq[
    [
        "SEQN",
        "PHQ9_TOTAL",
        "DEPRESSION_RISK"
    ]
].copy()


# Merge

for additional_df in [

    demo_small,
    bmx_small,
    slq_small,
    smq_small,
    paq_small,
    huq_small

]:

    model_df = model_df.merge(

        additional_df,

        on="SEQN",

        how="left"

    )


print(
    f"Merged dataset: "
    f"{model_df.shape}"
)


# ================================================================
# 17. KEEP COMPLETE PHQ CASES
# ================================================================

model_df = model_df.loc[
    model_df[
        "DEPRESSION_RISK"
    ].notna()
].copy()


model_df[
    "DEPRESSION_RISK"
] = model_df[
    "DEPRESSION_RISK"
].astype(int)


print(
    f"Modelling participants: "
    f"{len(model_df):,}"
)


# ================================================================
# 18. DEFINE PREDICTORS
# ================================================================
#
# IMPORTANT:
#
# PHQ-9 items and PHQ9_TOTAL are NOT predictors.
#
# This prevents target leakage.
# ================================================================


PREDICTORS = [

    "RIAGENDR",
    "RIDAGEYR",
    "RIDRETH1",
    "DMDEDUC2",
    "INDHHIN2",
    "DMDMARTL",

    "BMXBMI",
    "BMXWAIST",

    "SLD012",
    "SLD013",
    "SLQ050",

    "SMQ020",
    "SMQ040",

    "PAQ605",
    "PAQ620",

    "PAD615",
    "PAD630",

    "HUQ010",
    "HUQ030"

]


AVAILABLE_FEATURES = [

    col
    for col in PREDICTORS
    if col in model_df.columns

]


missing_features = [

    col
    for col in PREDICTORS
    if col not in model_df.columns

]


print("\n")
print("=" * 70)
print("PREDICTOR VALIDATION")
print("=" * 70)


print(
    f"Requested predictors: "
    f"{len(PREDICTORS)}"
)


print(
    f"Available predictors: "
    f"{len(AVAILABLE_FEATURES)}"
)


if missing_features:

    print(
        "\nUnavailable predictors:"
    )

    for col in missing_features:

        print(
            " -",
            col
        )


print(
    "\nAvailable predictors:"
)


for col in AVAILABLE_FEATURES:

    print(
        " ✓",
        col
    )


if len(
    AVAILABLE_FEATURES
) < 10:

    raise RuntimeError(
        "Too few predictors available."
    )


# ================================================================
# 19. CREATE X AND Y
# ================================================================

X = model_df[
    AVAILABLE_FEATURES
].copy()


y = model_df[
    "DEPRESSION_RISK"
].astype(int).copy()


# ================================================================
# 20. IDENTIFY FEATURE TYPES
# ================================================================

CATEGORICAL_FEATURES = [

    "RIAGENDR",
    "RIDRETH1",
    "DMDEDUC2",
    "INDHHIN2",
    "DMDMARTL",
    "SLQ050",
    "SMQ020",
    "SMQ040",
    "PAQ605",
    "PAQ620",
    "HUQ010",
    "HUQ030"

]


categorical_features = [

    col
    for col in AVAILABLE_FEATURES
    if col in CATEGORICAL_FEATURES

]


numeric_features = [

    col
    for col in AVAILABLE_FEATURES
    if col not in categorical_features

]


print("\n")
print("=" * 70)
print("FEATURE TYPES")
print("=" * 70)


print(
    "\nNumeric features:"
)


print(
    numeric_features
)


print(
    "\nCategorical features:"
)


print(
    categorical_features
)


# ================================================================
# 21. FIX PANDAS.NA
# ================================================================
#
# This specifically prevents:
#
# TypeError:
# boolean value of NA is ambiguous
#
# ================================================================


for col in categorical_features:

    X[col] = X[col].astype(object)

    X[col] = X[col].where(
        X[col].notna(),
        np.nan
    )


for col in numeric_features:

    X[col] = pd.to_numeric(
        X[col],
        errors="coerce"
    )


# ================================================================
# 22. VERIFY NO pd.NA
# ================================================================

for col in X.columns:

    if X[col].dtype == "object":

        X[col] = X[col].replace(
            {
                pd.NA: np.nan
            }
        )


# ================================================================
# 23. TRAIN / TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = (

    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=RANDOM_STATE,

        stratify=y

    )

)


print("\n")
print("=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)


print(
    f"Training participants: "
    f"{len(X_train):,}"
)


print(
    f"Testing participants: "
    f"{len(X_test):,}"
)


print(
    f"Training positives: "
    f"{int(y_train.sum()):,}"
)


print(
    f"Testing positives: "
    f"{int(y_test.sum()):,}"
)


# ================================================================
# 24. PREPROCESSOR
# ================================================================


numeric_pipeline = Pipeline(

    steps=[

        (
            "imputer",

            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",

            StandardScaler()
        )

    ]

)


categorical_pipeline = Pipeline(

    steps=[

        (
            "imputer",

            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "onehot",

            OneHotEncoder(

                handle_unknown="ignore",

                sparse_output=False

            )
        )

    ]

)


preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",

            numeric_pipeline,

            numeric_features

        ),

        (
            "categorical",

            categorical_pipeline,

            categorical_features

        )

    ],

    remainder="drop"

)


# ================================================================
# 25. PREPROCESS
# ================================================================

print("\n")
print("=" * 70)
print("PREPROCESSING")
print("=" * 70)


X_train_processed = (

    preprocessor
    .fit_transform(
        X_train
    )

)


X_test_processed = (

    preprocessor
    .transform(
        X_test
    )

)


feature_names = (

    preprocessor
    .get_feature_names_out()

)


X_train_processed = pd.DataFrame(

    X_train_processed,

    columns=feature_names,

    index=X_train.index

)


X_test_processed = pd.DataFrame(

    X_test_processed,

    columns=feature_names,

    index=X_test.index

)


print(
    "✓ Preprocessing completed."
)


print(
    f"Processed features: "
    f"{X_train_processed.shape[1]}"
)


# ================================================================
# 26. FINAL NUMERICAL VALIDATION
# ================================================================

if X_train_processed.isna().any().any():

    raise RuntimeError(
        "NaN values remain in training data."
    )


if X_test_processed.isna().any().any():

    raise RuntimeError(
        "NaN values remain in testing data."
    )


if not np.isfinite(
    X_train_processed.values
).all():

    raise RuntimeError(
        "Infinite values found in training data."
    )


if not np.isfinite(
    X_test_processed.values
).all():

    raise RuntimeError(
        "Infinite values found in testing data."
    )


print(
    "✓ No missing or infinite values remain."
)


# ================================================================
# 27. CLASS IMBALANCE
# ================================================================

negative_train = int(
    (y_train == 0).sum()
)


positive_train = int(
    (y_train == 1).sum()
)


scale_pos_weight = (

    negative_train
    /
    positive_train

)


print("\n")
print("=" * 70)
print("CLASS BALANCE")
print("=" * 70)


print(
    f"Negative training cases: "
    f"{negative_train:,}"
)


print(
    f"Positive training cases: "
    f"{positive_train:,}"
)


print(
    f"Scale positive weight: "
    f"{scale_pos_weight:.2f}"
)


# ================================================================
# 28. MODEL DEFINITIONS
# ================================================================


models = {

    "Logistic Regression":

        LogisticRegression(

            max_iter=3000,

            class_weight="balanced",

            random_state=RANDOM_STATE

        ),


    "Random Forest":

        RandomForestClassifier(

            n_estimators=500,

            max_depth=12,

            min_samples_leaf=5,

            class_weight="balanced_subsample",

            random_state=RANDOM_STATE,

            n_jobs=-1

        ),


    "XGBoost":

        xgb.XGBClassifier(

            n_estimators=500,

            max_depth=4,

            learning_rate=0.03,

            subsample=0.85,

            colsample_bytree=0.85,

            objective="binary:logistic",

            eval_metric="logloss",

            scale_pos_weight=scale_pos_weight,

            random_state=RANDOM_STATE,

            n_jobs=-1

        )

}


# ================================================================
# 29. THRESHOLD OPTIMISATION
# ================================================================
#
# The default 0.5 threshold is not ideal for an imbalanced
# screening problem.
#
# The threshold is selected ONLY using the training data.
#
# The test set remains untouched until final evaluation.
# ================================================================


def optimise_threshold(

    y_true,

    probabilities

):

    fpr, tpr, thresholds = roc_curve(

        y_true,

        probabilities

    )


    finite = np.isfinite(
        thresholds
    )


    fpr = fpr[finite]

    tpr = tpr[finite]

    thresholds = thresholds[finite]


    youden_j = (

        tpr
        -
        fpr

    )


    best_index = np.argmax(
        youden_j
    )


    return float(
        thresholds[
            best_index
        ]
    )


# ================================================================
# 30. MODEL EVALUATION FUNCTION
# ================================================================


def evaluate_model(

    y_true,

    probabilities,

    threshold

):

    predictions = (

        probabilities
        >= threshold

    ).astype(int)


    tn, fp, fn, tp = (

        confusion_matrix(

            y_true,

            predictions,

            labels=[0, 1]

        ).ravel()

    )


    sensitivity = (

        tp
        /
        (tp + fn)

        if tp + fn > 0

        else np.nan

    )


    specificity = (

        tn
        /
        (tn + fp)

        if tn + fp > 0

        else np.nan

    )


    ppv = (

        tp
        /
        (tp + fp)

        if tp + fp > 0

        else np.nan

    )


    npv = (

        tn
        /
        (tn + fn)

        if tn + fn > 0

        else np.nan

    )


    return {

        "Threshold":
            threshold,

        "Accuracy":
            accuracy_score(
                y_true,
                predictions
            ),

        "Precision_PPV":
            ppv,

        "Sensitivity_TPR":
            sensitivity,

        "Specificity":
            specificity,

        "F1":
            f1_score(
                y_true,
                predictions,
                zero_division=0
            ),

        "ROC_AUC":
            roc_auc_score(
                y_true,
                probabilities
            ),

        "PR_AUC":
            average_precision_score(
                y_true,
                probabilities
            ),

        "Brier_Score":
            brier_score_loss(
                y_true,
                probabilities
            ),

        "NPV":
            npv,

        "TN":
            tn,

        "FP":
            fp,

        "FN":
            fn,

        "TP":
            tp

    }


# ================================================================
# 31. TRAIN MODELS
# ================================================================

print("\n")
print("=" * 70)
print("TRAINING MACHINE LEARNING MODELS")
print("=" * 70)


trained_models = {}

test_probabilities = {}

model_thresholds = {}

results = []


for model_name, model in models.items():

    print(
        f"\nTraining {model_name}..."
    )


    model.fit(

        X_train_processed,

        y_train

    )


    train_probability = (

        model
        .predict_proba(
            X_train_processed
        )[:, 1]

    )


    test_probability = (

        model
        .predict_proba(
            X_test_processed
        )[:, 1]

    )


    threshold = optimise_threshold(

        y_train,

        train_probability

    )


    metrics = evaluate_model(

        y_test,

        test_probability,

        threshold

    )


    metrics[
        "Model"
    ] = model_name


    results.append(
        metrics
    )


    trained_models[
        model_name
    ] = model


    test_probabilities[
        model_name
    ] = test_probability


    model_thresholds[
        model_name
    ] = threshold


    print(
        f"  Threshold: "
        f"{threshold:.4f}"
    )

    print(
        f"  ROC-AUC: "
        f"{metrics['ROC_AUC']:.4f}"
    )

    print(
        f"  PR-AUC: "
        f"{metrics['PR_AUC']:.4f}"
    )

    print(
        f"  Sensitivity: "
        f"{metrics['Sensitivity_TPR']:.4f}"
    )

    print(
        f"  Specificity: "
        f"{metrics['Specificity']:.4f}"
    )

    print(
        f"  F1: "
        f"{metrics['F1']:.4f}"
    )


# ================================================================
# 32. MODEL RESULTS TABLE
# ================================================================


results_df = pd.DataFrame(
    results
)


results_df = results_df[
    [

        "Model",

        "Threshold",

        "Accuracy",

        "Precision_PPV",

        "Sensitivity_TPR",

        "Specificity",

        "F1",

        "ROC_AUC",

        "PR_AUC",

        "Brier_Score",

        "NPV",

        "TN",

        "FP",

        "FN",

        "TP"

    ]
]


print("\n")
print("=" * 70)
print("MODEL PERFORMANCE")
print("=" * 70)


display(
    results_df.round(4)
)


results_df.to_csv(

    f"{OUTPUT_DIR}/model_results.csv",

    index=False

)


# ================================================================
# 33. SELECT BEST MODEL
# ================================================================


best_model_name = (

    results_df

    .sort_values(

        "ROC_AUC",

        ascending=False

    )

    .iloc[0][
        "Model"
    ]

)


best_model = (

    trained_models[
        best_model_name
    ]

)


best_probability = (

    test_probabilities[
        best_model_name
    ]

)


best_threshold = (

    model_thresholds[
        best_model_name
    ]

)


best_result = (

    results_df[

        results_df[
            "Model"
        ]
        ==
        best_model_name

    ]

    .iloc[0]

)


print("\n")
print("=" * 70)
print("BEST MODEL")
print("=" * 70)


print(
    f"Best model: "
    f"{best_model_name}"
)


print(
    f"Threshold: "
    f"{best_threshold:.4f}"
)


print(
    f"ROC-AUC: "
    f"{best_result['ROC_AUC']:.4f}"
)


# ================================================================
# 34. CONFUSION MATRIX
# ================================================================


best_prediction = (

    best_probability
    >= best_threshold

).astype(int)


cm = confusion_matrix(

    y_test,

    best_prediction,

    labels=[0, 1]

)


plt.figure(
    figsize=(7, 6)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "Predicted Low",
        "Predicted Elevated"
    ],

    yticklabels=[
        "Actual Low",
        "Actual Elevated"
    ]

)


plt.title(
    f"Confusion Matrix - {best_model_name}"
)

plt.xlabel(
    "Prediction"
)

plt.ylabel(
    "Actual"
)

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/confusion_matrix.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ================================================================
# 35. ROC CURVE
# ================================================================


plt.figure(
    figsize=(8, 6)
)


for model_name, probability in (

    test_probabilities.items()

):

    fpr, tpr, _ = roc_curve(

        y_test,

        probability

    )


    auc = roc_auc_score(

        y_test,

        probability

    )


    plt.plot(

        fpr,

        tpr,

        label=(
            f"{model_name} "
            f"(AUC={auc:.3f})"
        )

    )


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve Comparison"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/roc_curve.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ================================================================
# 36. PRECISION-RECALL CURVE
# ================================================================


plt.figure(
    figsize=(8, 6)
)


for model_name, probability in (

    test_probabilities.items()

):

    precision, recall, _ = (

        precision_recall_curve(

            y_test,

            probability

        )

    )


    ap = average_precision_score(

        y_test,

        probability

    )


    plt.plot(

        recall,

        precision,

        label=(
            f"{model_name} "
            f"(AP={ap:.3f})"
        )

    )


plt.xlabel(
    "Recall / Sensitivity"
)

plt.ylabel(
    "Precision / PPV"
)

plt.title(
    "Precision-Recall Curve"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/precision_recall_curve.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ================================================================
# 37. CALIBRATION CURVE
# ================================================================


plt.figure(
    figsize=(8, 6)
)


for model_name, probability in (

    test_probabilities.items()

):

    fraction_positive, mean_prediction = (

        calibration_curve(

            y_test,

            probability,

            n_bins=10,

            strategy="quantile"

        )

    )


    plt.plot(

        mean_prediction,

        fraction_positive,

        marker="o",

        label=model_name

    )


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "Mean predicted probability"
)

plt.ylabel(
    "Observed frequency"
)

plt.title(
    "Model Calibration"
)

plt.legend()

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/calibration_curve.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ================================================================
# 38. SHAP EXPLAINABILITY
# ================================================================


print("\n")
print("=" * 70)
print("SHAP EXPLAINABILITY")
print("=" * 70)


# Use XGBoost specifically because SHAP TreeExplainer provides
# efficient and interpretable explanations for tree models.

xgb_model = trained_models[
    "XGBoost"
]


shap_sample_size = min(

    500,

    len(X_test_processed)

)


shap_sample = (

    X_test_processed
    .iloc[
        :shap_sample_size
    ]
)


shap_explainer = shap.TreeExplainer(
    xgb_model
)


shap_values = shap_explainer(
    shap_sample
)


print(
    f"SHAP observations: "
    f"{len(shap_sample):,}"
)


# ================================================================
# 39. SHAP SUMMARY PLOT
# ================================================================


shap.summary_plot(

    shap_values.values,

    shap_sample,

    feature_names=feature_names,

    show=False

)


plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/shap_summary.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ================================================================
# 40. SHAP GLOBAL IMPORTANCE
# ================================================================


mean_abs_shap = (

    np.abs(
        shap_values.values
    )
    .mean(axis=0)

)


shap_importance = pd.DataFrame({

    "Feature":
        feature_names,

    "Mean_Absolute_SHAP":
        mean_abs_shap

})


shap_importance = (

    shap_importance

    .sort_values(

        "Mean_Absolute_SHAP",

        ascending=False

    )

)


print("\n")
print(
    "TOP 20 SHAP FEATURES"
)


display(

    shap_importance
    .head(20)
    .round(6)

)


shap_importance.to_csv(

    f"{OUTPUT_DIR}/shap_importance.csv",

    index=False

)


# ================================================================
# 41. SHAP WATERFALL
# ================================================================


shap.plots.waterfall(

    shap_values[0],

    max_display=15

)


plt.savefig(

    f"{OUTPUT_DIR}/shap_waterfall.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ================================================================
# 42. LIME EXPLAINABILITY
# ================================================================


print("\n")
print("=" * 70)
print("LIME EXPLAINABILITY")
print("=" * 70)


lime_explainer = LimeTabularExplainer(

    X_train_processed.values,

    feature_names=list(
        feature_names
    ),

    class_names=[
        "Low risk",
        "Elevated risk"
    ],

    mode="classification",

    discretize_continuous=True,

    random_state=RANDOM_STATE

)


lime_instance = (

    X_test_processed
    .iloc[0]
    .values

)


lime_explanation = (

    lime_explainer

    .explain_instance(

        lime_instance,

        xgb_model.predict_proba,

        num_features=15,

        num_samples=5000

    )

)


lime_df = pd.DataFrame(

    lime_explanation.as_list(),

    columns=[
        "Feature",
        "LIME_Weight"
    ]

)


print(
    "\nLIME explanation:"
)


display(
    lime_df
)


lime_df.to_csv(

    f"{OUTPUT_DIR}/lime_explanation.csv",

    index=False

)


# ================================================================
# 43. CORRECTED SHAP / LIME FEATURE AGREEMENT
# ================================================================
#
# IMPORTANT:
#
# LIME returns feature CONDITIONS such as:
#
#   numeric__BMXWAIST <= 0.25
#
#   -0.69 < numeric__BMXWAIST <= -0.07
#
# SHAP returns the underlying feature name:
#
#   numeric__BMXWAIST
#
# Therefore, comparing the raw strings produces an artificially
# low/zero Jaccard score.
#
# This section converts LIME conditions into their underlying
# feature names before calculating agreement.
# ================================================================


def canonicalize_lime_feature(
    lime_feature,
    known_features
):

    # First try exact feature matching

    if lime_feature in known_features:

        return lime_feature


    # Find the longest matching feature name contained inside
    # the LIME condition.

    matches = [

        feature

        for feature in known_features

        if feature in lime_feature

    ]


    if matches:

        return max(
            matches,
            key=len
        )


    return lime_feature


# ------------------------------------------------
# SHAP top 10
# ------------------------------------------------

shap_top_features = list(

    shap_importance

    .head(10)[
        "Feature"
    ]

)


# ------------------------------------------------
# LIME top 10
# ------------------------------------------------

lime_ranked = (

    lime_df

    .assign(

        Absolute_LIME_Weight=

        lambda df:

        np.abs(
            df[
                "LIME_Weight"
            ]
        )

    )

    .sort_values(

        "Absolute_LIME_Weight",

        ascending=False

    )

)


lime_top_raw = (

    lime_ranked

    .head(10)[
        "Feature"
    ]

)


# Convert LIME conditions to canonical feature names

lime_top_features = [

    canonicalize_lime_feature(

        feature,

        set(feature_names)

    )

    for feature in lime_top_raw

]


# ------------------------------------------------
# Convert to sets
# ------------------------------------------------

shap_set = set(
    shap_top_features
)


lime_set = set(
    lime_top_features
)


# ------------------------------------------------
# Calculate intersection and union
# ------------------------------------------------

intersection = (

    shap_set
    &
    lime_set

)


union = (

    shap_set
    |
    lime_set

)


if len(union) > 0:

    shap_lime_jaccard = (

        len(intersection)
        /
        len(union)

    )

else:

    shap_lime_jaccard = 0.0


# ------------------------------------------------
# Display results
# ------------------------------------------------

print("\n")
print("=" * 70)
print("CORRECTED SHAP / LIME AGREEMENT")
print("=" * 70)


print(
    "\nSHAP top 10:"
)


for feature in shap_top_features:

    print(
        "  ",
        feature
    )


print(
    "\nLIME top 10 - raw:"
)


for feature in lime_top_raw:

    print(
        "  ",
        feature
    )


print(
    "\nLIME top 10 - canonical:"
)


for feature in lime_top_features:

    print(
        "  ",
        feature
    )


print(
    "\nShared features:"
)


for feature in sorted(
    intersection
):

    print(
        "  ✓",
        feature
    )


print(
    f"\nNumber of shared features: "
    f"{len(intersection)}"
)


print(
    f"SHAP features in comparison: "
    f"{len(shap_set)}"
)


print(
    f"LIME features in comparison: "
    f"{len(lime_set)}"
)


print(
    f"\nCorrected Jaccard agreement: "
    f"{shap_lime_jaccard:.4f}"
)


# ------------------------------------------------
# Save agreement results
# ------------------------------------------------

agreement_summary = pd.DataFrame({

    "Metric": [

        "SHAP_LIME_Jaccard",

        "Shared_Features",

        "SHAP_Top_Features",

        "LIME_Top_Features"

    ],

    "Value": [

        shap_lime_jaccard,

        len(intersection),

        len(shap_set),

        len(lime_set)

    ]

})


agreement_summary.to_csv(

    f"{OUTPUT_DIR}/shap_lime_agreement.csv",

    index=False

)


print(
    "\n✓ Corrected SHAP/LIME agreement saved."
)

# ================================================================
# 44. LIME STABILITY
# ================================================================
#
# LIME is stochastic.
#
# We therefore run it multiple times and calculate the average
# Jaccard similarity between explanation feature sets.
# ================================================================


def generate_lime_features(
    seed
):

    explainer = LimeTabularExplainer(

        X_train_processed.values,

        feature_names=list(
            feature_names
        ),

        class_names=[
            "Low risk",
            "Elevated risk"
        ],

        mode="classification",

        discretize_continuous=True,

        random_state=seed

    )


    explanation = (

        explainer

        .explain_instance(

            lime_instance,

            xgb_model.predict_proba,

            num_features=10,

            num_samples=3000

        )

    )


    return set(

        feature

        for feature, weight
        in explanation.as_list()

    )


lime_sets = []


for seed in [
    1,
    2,
    3,
    4,
    5
]:

    lime_sets.append(

        generate_lime_features(
            seed
        )

    )


pairwise_jaccards = []


for i in range(
    len(lime_sets)
):

    for j in range(

        i + 1,

        len(lime_sets)

    ):

        a = lime_sets[i]

        b = lime_sets[j]


        union_ab = a | b


        intersection_ab = a & b


        if len(union_ab) > 0:

            pairwise_jaccards.append(

                len(
                    intersection_ab
                )
                /
                len(
                    union_ab
                )

            )


lime_stability = (

    np.mean(
        pairwise_jaccards
    )

)


print(
    f"LIME stability: "
    f"{lime_stability:.4f}"
)


# ================================================================
# 45. FAIRNESS DATASET
# ================================================================


fairness_df = (

    model_df
    .loc[
        X_test.index
    ]
    .copy()

)


fairness_df[
    "Predicted_Probability"
] = best_probability


fairness_df[
    "Predicted_Risk"
] = (

    best_probability
    >= best_threshold

).astype(int)


# ================================================================
# 46. AGE GROUPS
# ================================================================


fairness_df[
    "Age_Group"
] = pd.cut(

    fairness_df[
        "RIDAGEYR"
    ],

    bins=[
        17,
        29,
        44,
        59,
        120
    ],

    labels=[
        "18-29",
        "30-44",
        "45-59",
        "60+"
    ]

)


# ================================================================
# 47. FAIRNESS METRICS FUNCTION
# ================================================================


def calculate_subgroup_metrics(
    subgroup
):

    y_true = subgroup[
        "DEPRESSION_RISK"
    ].astype(int).values


    y_pred = subgroup[
        "Predicted_Risk"
    ].astype(int).values


    probability = subgroup[
        "Predicted_Probability"
    ].values


    tn, fp, fn, tp = (

        confusion_matrix(

            y_true,

            y_pred,

            labels=[0, 1]

        ).ravel()

    )


    sensitivity = (

        tp
        /
        (tp + fn)

        if tp + fn > 0

        else np.nan

    )


    specificity = (

        tn
        /
        (tn + fp)

        if tn + fp > 0

        else np.nan

    )


    fpr = (

        fp
        /
        (fp + tn)

        if fp + tn > 0

        else np.nan

    )


    fnr = (

        fn
        /
        (fn + tp)

        if fn + tp > 0

        else np.nan

    )


    ppv = (

        tp
        /
        (tp + fp)

        if tp + fp > 0

        else np.nan

    )


    npv = (

        tn
        /
        (tn + fn)

        if tn + fn > 0

        else np.nan

    )


    if len(
        np.unique(y_true)
    ) == 2:

        auc = roc_auc_score(

            y_true,

            probability

        )

    else:

        auc = np.nan


    return {

        "N":
            len(subgroup),

        "Positive_N":
            int(y_true.sum()),

        "Prevalence":
            float(
                np.mean(y_true)
            ),

        "Selection_Rate":
            float(
                np.mean(y_pred)
            ),

        "Sensitivity_TPR":
            sensitivity,

        "Specificity":
            specificity,

        "FPR":
            fpr,

        "FNR":
            fnr,

        "PPV":
            ppv,

        "NPV":
            npv,

        "ROC_AUC":
            auc,

        "Mean_Predicted_Risk":
            float(
                np.mean(
                    probability
                )
            )

    }


# ================================================================
# 48. FAIRNESS ANALYSIS
# ================================================================


FAIRNESS_VARIABLES = {

    "Sex":
        "RIAGENDR",

    "Race_Ethnicity":
        "RIDRETH1",

    "Age_Group":
        "Age_Group"

}


fairness_records = []


for group_type, variable in (

    FAIRNESS_VARIABLES.items()

):

    for group_value, subgroup in (

        fairness_df
        .groupby(
            variable,
            dropna=False
        )

    ):

        metrics = (

            calculate_subgroup_metrics(
                subgroup
            )

        )


        metrics[
            "Group_Type"
        ] = group_type


        metrics[
            "Group"
        ] = str(
            group_value
        )


        fairness_records.append(
            metrics
        )


fairness_results = pd.DataFrame(
    fairness_records
)


fairness_results = fairness_results[

    [

        "Group_Type",

        "Group",

        "N",

        "Positive_N",

        "Prevalence",

        "Selection_Rate",

        "Sensitivity_TPR",

        "Specificity",

        "FPR",

        "FNR",

        "PPV",

        "NPV",

        "ROC_AUC",

        "Mean_Predicted_Risk"

    ]

]


print("\n")
print("=" * 70)
print("FAIRNESS ANALYSIS")
print("=" * 70)


display(
    fairness_results.round(4)
)


fairness_results.to_csv(

    f"{OUTPUT_DIR}/fairness_results.csv",

    index=False

)


# ================================================================
# 49. FAIRNESS GAPS
# ================================================================


fairness_gap_records = []


for group_type in (

    fairness_results[
        "Group_Type"
    ].unique()

):

    subset = fairness_results[

        fairness_results[
            "Group_Type"
        ]
        ==
        group_type

    ]


    tpr = subset[
        "Sensitivity_TPR"
    ].dropna()


    fpr = subset[
        "FPR"
    ].dropna()


    selection = subset[
        "Selection_Rate"
    ].dropna()


    fairness_gap_records.append({

        "Group_Type":
            group_type,

        "Equal_Opportunity_TPR_Gap":
            (
                tpr.max()
                -
                tpr.min()
                if len(tpr)
                else np.nan
            ),

        "FPR_Gap":
            (
                fpr.max()
                -
                fpr.min()
                if len(fpr)
                else np.nan
            ),

        "Demographic_Parity_Selection_Gap":
            (
                selection.max()
                -
                selection.min()
                if len(selection)
                else np.nan
            )

    })


fairness_gaps = pd.DataFrame(
    fairness_gap_records
)


print(
    "\nFairness gaps:"
)


display(
    fairness_gaps.round(4)
)


fairness_gaps.to_csv(

    f"{OUTPUT_DIR}/fairness_gaps.csv",

    index=False

)


# ================================================================
# 50. MODEL FEATURE IMPORTANCE
# ================================================================


if hasattr(
    xgb_model,
    "feature_importances_"
):

    xgb_importance = pd.DataFrame({

        "Feature":
            feature_names,

        "Importance":
            xgb_model.feature_importances_

    })


    xgb_importance = (

        xgb_importance

        .sort_values(

            "Importance",

            ascending=False

        )

    )


    print("\n")
    print("=" * 70)
    print("XGBOOST FEATURE IMPORTANCE")
    print("=" * 70)


    display(

        xgb_importance
        .head(20)
        .round(6)

    )


    xgb_importance.to_csv(

        f"{OUTPUT_DIR}/xgboost_feature_importance.csv",

        index=False

    )


# ================================================================
# 51. SAVE PREPROCESSOR AND MODELS
# ================================================================

print("\n")
print("=" * 70)
print("SAVING MODELS")
print("=" * 70)


# ------------------------------------------------
# Save the preprocessing pipeline
# ------------------------------------------------

joblib.dump(
    preprocessor,
    f"{OUTPUT_DIR}/preprocessor.joblib"
)


# ------------------------------------------------
# Save XGBoost model
# ------------------------------------------------

joblib.dump(
    xgb_model,
    f"{OUTPUT_DIR}/xgboost_model.joblib"
)


# ------------------------------------------------
# Save the selected/best model
# ------------------------------------------------

joblib.dump(
    best_model,
    f"{OUTPUT_DIR}/best_model.joblib"
)


# ------------------------------------------------
# DO NOT pickle lime_explainer
#
# LIME contains an internal lambda function that
# cannot be serialized reliably with Python 3.13.
#
# Instead, save the information required to recreate
# the explainer later.
# ------------------------------------------------

lime_metadata = {

    "feature_names":
        list(feature_names),

    "class_names":
        [
            "Low risk",
            "Elevated risk"
        ],

    "mode":
        "classification",

    "discretize_continuous":
        True,

    "random_state":
        RANDOM_STATE,

    "num_features":
        15,

    "num_samples":
        5000

}


joblib.dump(

    lime_metadata,

    f"{OUTPUT_DIR}/lime_metadata.joblib"

)


# ------------------------------------------------
# Save feature names
# ------------------------------------------------

pd.DataFrame({

    "Feature":
        feature_names

}).to_csv(

    f"{OUTPUT_DIR}/processed_feature_names.csv",

    index=False

)


print("\n✓ Preprocessor saved")

print("✓ XGBoost model saved")

print("✓ Best model saved")

print("✓ LIME metadata saved")

print(
    "\n✓ LIME explainer itself was NOT pickled "
    "because Python 3.13/LIME cannot serialize "
    "its internal lambda function."
)


# ================================================================
# HOW TO RECREATE LIME LATER
# ================================================================
#
# The following code can recreate the LIME explainer whenever
# needed instead of loading a pickle.
# ================================================================

def create_lime_explainer():

    return LimeTabularExplainer(

        X_train_processed.values,

        feature_names=list(
            feature_names
        ),

        class_names=[
            "Low risk",
            "Elevated risk"
        ],

        mode="classification",

        discretize_continuous=True,

        random_state=RANDOM_STATE

    )


# Test that LIME can be recreated successfully

lime_explainer_test = create_lime_explainer()


print(
    "✓ LIME explainer can be recreated successfully."
)


# ================================================================
# 52. SAVE MODELLING DATASET
# ================================================================

model_df.to_csv(

    f"{OUTPUT_DIR}/modelling_dataset.csv",

    index=False

)


print(
    "✓ Modelling dataset saved"
)


# ================================================================
# 53. SAVE TEST PREDICTIONS
# ================================================================

predictions_df = pd.DataFrame({

    "SEQN":
        model_df.loc[
            X_test.index,
            "SEQN"
        ].values,

    "Actual_Risk":
        y_test.values,

    "Predicted_Probability":
        best_probability,

    "Predicted_Risk":
        best_prediction

})


predictions_df.to_csv(

    f"{OUTPUT_DIR}/test_predictions.csv",

    index=False

)


print(
    "✓ Test predictions saved"
)


# ================================================================
# 54. SAVE SUMMARY JSON
# ================================================================

summary = {

    "Dataset":
        "CDC NHANES 2017-2018",

    "Target":
        "PHQ-9 >= 10",

    "Complete_PHQ9_Cases":
        int(complete_phq),

    "Positive_Cases":
        int(positive_cases),

    "Negative_Cases":
        int(negative_cases),

    "Positive_Prevalence_Percent":
        float(prevalence),

    "Modelling_Sample":
        int(len(model_df)),

    "Training_Sample":
        int(len(X_train)),

    "Testing_Sample":
        int(len(X_test)),

    "Processed_Features":
        int(
            X_train_processed.shape[1]
        ),

    "Best_Model":
        best_model_name,

    "Threshold":
        float(best_threshold),

    "Accuracy":
        float(
            best_result[
                "Accuracy"
            ]
        ),

    "Precision_PPV":
        float(
            best_result[
                "Precision_PPV"
            ]
        ),

    "Sensitivity":
        float(
            best_result[
                "Sensitivity_TPR"
            ]
        ),

    "Specificity":
        float(
            best_result[
                "Specificity"
            ]
        ),

    "F1":
        float(
            best_result[
                "F1"
            ]
        ),

    "ROC_AUC":
        float(
            best_result[
                "ROC_AUC"
            ]
        ),

    "PR_AUC":
        float(
            best_result[
                "PR_AUC"
            ]
        ),

    "Brier_Score":
        float(
            best_result[
                "Brier_Score"
            ]
        ),

    "NPV":
        float(
            best_result[
                "NPV"
            ]
        ),

    "SHAP_LIME_Jaccard":
        float(
            shap_lime_jaccard
        ),

    "LIME_Stability":
        float(
            lime_stability
        )

}


with open(

    f"{OUTPUT_DIR}/summary.json",

    "w"

) as f:

    json.dump(

        summary,

        f,

        indent=4

    )


print(
    "✓ Summary JSON saved"
)


# ================================================================
# 55. FINAL CLINICIAN-FACING EXAMPLE
# ================================================================

print("\n")
print("=" * 70)
print("CLINICIAN-FACING EXAMPLE")
print("=" * 70)


example_original = X_test.iloc[
    [0]
]


example_processed = (

    preprocessor
    .transform(
        example_original
    )

)


example_probability = (

    xgb_model
    .predict_proba(
        example_processed
    )[:, 1][0]

)


example_prediction = (

    example_probability
    >= best_threshold

)


print(
    f"Predicted risk probability: "
    f"{example_probability:.4f}"
)


print(
    f"Predicted risk percentage: "
    f"{example_probability * 100:.2f}%"
)


print(
    f"Decision threshold: "
    f"{best_threshold:.4f}"
)


if example_prediction:

    print(
        "\nMODEL OUTPUT:"
    )

    print(
        "ELEVATED DEPRESSIVE SYMPTOM RISK"
    )

else:

    print(
        "\nMODEL OUTPUT:"
    )

    print(
        "BELOW MODEL SCREENING THRESHOLD"
    )


print(
    "\nIMPORTANT:"
)


print(
    "This is an AI decision-support estimate,"
)


print(
    "NOT a diagnosis of depression."
)


# ================================================================
# 56. LIST OUTPUT FILES
# ================================================================

print("\n")
print("=" * 70)
print("OUTPUT FILES")
print("=" * 70)


for filename in sorted(

    os.listdir(
        OUTPUT_DIR
    )

):

    filepath = os.path.join(

        OUTPUT_DIR,

        filename

    )


    size_kb = (

        os.path.getsize(
            filepath
        )
        /
        1024

    )


    print(

        f"✓ {filename:<45} "
        f"{size_kb:,.1f} KB"

    )


# ================================================================
# 57. FINAL RESULTS
# ================================================================

print("\n\n")
print("=" * 70)
print("PIPELINE COMPLETED SUCCESSFULLY")
print("=" * 70)


print("\nDATASET")

print(
    "CDC NHANES 2017-2018"
)


print("\nTARGET")

print(
    "Elevated depressive symptom risk "
    "(PHQ-9 >= 10)"
)


print("\nSAMPLE")

print(
    f"Complete PHQ-9 cases: "
    f"{complete_phq:,}"
)


print(
    f"Positive cases: "
    f"{positive_cases:,}"
)


print(
    f"Prevalence: "
    f"{prevalence:.2f}%"
)


print(
    f"Modelling sample: "
    f"{len(model_df):,}"
)


print(
    f"Processed features: "
    f"{X_train_processed.shape[1]}"
)


print("\nBEST MODEL")

print(
    best_model_name
)


print(
    f"Threshold: "
    f"{best_threshold:.4f}"
)


print(
    f"ROC-AUC: "
    f"{best_result['ROC_AUC']:.4f}"
)


print(
    f"PR-AUC: "
    f"{best_result['PR_AUC']:.4f}"
)


print(
    f"Sensitivity: "
    f"{best_result['Sensitivity_TPR']:.4f}"
)


print(
    f"Specificity: "
    f"{best_result['Specificity']:.4f}"
)


print(
    f"Precision / PPV: "
    f"{best_result['Precision_PPV']:.4f}"
)


print(
    f"F1: "
    f"{best_result['F1']:.4f}"
)


print(
    f"Brier score: "
    f"{best_result['Brier_Score']:.4f}"
)


print("\nXAI")

print(
    f"SHAP/LIME Jaccard: "
    f"{shap_lime_jaccard:.4f}"
)


print(
    f"LIME stability: "
    f"{lime_stability:.4f}"
)


print("\n")
print("=" * 70)
print("RESEARCH INTERPRETATION")
print("=" * 70)


print("""
The prototype predicts elevated depressive symptom risk,
defined operationally as PHQ-9 total score >=10.

The PHQ-9 questionnaire itself is NOT included as a predictor.
This prevents target leakage.

The system estimates elevated depressive symptom risk from
demographic, anthropometric, sleep, smoking, physical activity
and healthcare-utilisation variables.

The dataset is NHANES rather than NHS primary-care EHR data.

Therefore, this work should be described as an XAI
methodological proof-of-concept rather than a clinically
validated NHS deployment.

External validation using representative NHS primary-care
data would be required before clinical deployment.

The model output should never be interpreted as a diagnosis.
""")


print("=" * 70)

print(
    "ALL DONE"
)

print("=" * 70)
```

# **preprocessing.py**

```python
%%writefile /content/xai_mental_health_model/backend/preprocessing.py

import os
import joblib
import pandas as pd


PROJECT_DIR = "/content/xai_mental_health_model"

PREPROCESSOR_PATH = os.path.join(
    PROJECT_DIR,
    "preprocessor.joblib"
)

# Load the EXACT preprocessing pipeline used during model training
preprocessor = joblib.load(PREPROCESSOR_PATH)


def get_required_columns():
    """
    Return the exact raw variables expected by the saved
    dissertation preprocessing pipeline.
    """

    columns = []

    for _, _, transformer_columns in preprocessor.transformers_:
        columns.extend(list(transformer_columns))

    return columns


def validate_input(data):
    """
    Validate incoming assessment data.

    This performs API-level validation while ensuring that
    all variables required by the saved training pipeline
    are present.
    """

    if not isinstance(data, dict):
        raise ValueError(
            "Input must be a JSON object."
        )

    # --------------------------------------------------
    # Check the exact variables required by the model
    # --------------------------------------------------

    required_columns = get_required_columns()

    missing = [
        column
        for column in required_columns
        if column not in data
    ]

    if missing:
        raise ValueError(
            "Missing required variables: "
            + ", ".join(missing)
        )

    # --------------------------------------------------
    # Basic validation of selected numerical variables
    # --------------------------------------------------

    try:
        age = float(data["RIDAGEYR"])
        bmi = float(data["BMXBMI"])
        sleep_weekday = float(data["SLD012"])
        sleep_weekend = float(data["SLD013"])
    except (TypeError, ValueError):
        raise ValueError(
            "Age, BMI and sleep variables must be numeric."
        )

    if age < 18 or age > 120:
        raise ValueError(
            "Age must be between 18 and 120."
        )

    if bmi <= 0 or bmi > 100:
        raise ValueError(
            "BMI value is outside the permitted range."
        )

    if sleep_weekday < 0 or sleep_weekday > 24:
        raise ValueError(
            "Weekday sleep duration must be between 0 and 24 hours."
        )

    if sleep_weekend < 0 or sleep_weekend > 24:
        raise ValueError(
            "Weekend sleep duration must be between 0 and 24 hours."
        )

    return True


def prepare_input(data):
    """
    Convert raw assessment input into the exact 69-feature
    representation expected by the trained model.

    The saved preprocessing pipeline is used directly.
    No alternative encoding or fallback preprocessing is used.
    """

    validate_input(data)

    required_columns = get_required_columns()

    df = pd.DataFrame(
        [data],
        columns=required_columns
    )

    return preprocessor.transform(df)
```

# **predict.py**

```python
%%writefile /content/xai_mental_health_model/backend/predict.py

import os
import joblib

PROJECT_DIR = "/content/xai_mental_health_model"

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "best_model.joblib"
)

model = joblib.load(MODEL_PATH)

# Threshold used in the dissertation results
THRESHOLD = 0.4248


def predict_risk(X):

    probability = float(
        model.predict_proba(X)[0][1]
    )

    classification = (
        "Elevated predicted risk"
        if probability >= THRESHOLD
        else "Lower predicted risk"
    )

    return {
        "probability": round(probability, 6),
        "percentage": round(probability * 100, 2),
        "threshold": THRESHOLD,
        "classification": classification
    }
```

## **explain.py**

```python
%%writefile /content/xai_mental_health_model/backend/explain.py

import os
import joblib
import numpy as np
import pandas as pd
import shap

PROJECT_DIR = "/content/xai_mental_health_model"

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "xgboost_model.joblib"
)

FEATURE_PATH = os.path.join(
    PROJECT_DIR,
    "processed_feature_names.csv"
)

model = joblib.load(MODEL_PATH)

feature_table = pd.read_csv(FEATURE_PATH)

FEATURE_NAMES = (
    feature_table["Feature"]
    .astype(str)
    .tolist()
)


def explain_with_shap(X):

    explainer = shap.TreeExplainer(model)

    shap_values = explainer.shap_values(X)

    shap_values = np.asarray(shap_values)

    if shap_values.ndim == 3:
        shap_values = shap_values[0, :, 1]

    elif shap_values.ndim == 2:
        shap_values = shap_values[0]

    else:
        shap_values = shap_values.flatten()

    results = []

    for feature, value in zip(
        FEATURE_NAMES,
        shap_values
    ):

        value = float(value)

        results.append({
            "feature": feature,
            "contribution": round(value, 6),
            "direction": (
                "increases predicted risk"
                if value > 0
                else "decreases predicted risk"
            )
        })

    results.sort(
        key=lambda x: abs(x["contribution"]),
        reverse=True
    )

    return results
```

## **fairness.py**

```python
import numpy as np

from sklearn.metrics import (
    confusion_matrix,
    roc_auc_score
)


def calculate_group_metrics(
    y_true,
    y_probability,
    threshold=0.5
):

    y_true = np.asarray(y_true)
    y_probability = np.asarray(y_probability)

    y_pred = (
        y_probability >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else np.nan
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else np.nan
    )

    fnr = (
        fn / (fn + tp)
        if (fn + tp) > 0
        else np.nan
    )

    ppv = (
        tp / (tp + fp)
        if (tp + fp) > 0
        else np.nan
    )

    npv = (
        tn / (tn + fn)
        if (tn + fn) > 0
        else np.nan
    )

    try:
        auc = roc_auc_score(
            y_true,
            y_probability
        )
    except ValueError:
        auc = np.nan

    selection_rate = np.mean(y_pred)

    mean_predicted_risk = np.mean(
        y_probability
    )

    return {
        "sensitivity": float(sensitivity)
            if not np.isnan(sensitivity)
            else None,

        "specificity": float(specificity)
            if not np.isnan(specificity)
            else None,

        "fpr": float(fpr)
            if not np.isnan(fpr)
            else None,

        "fnr": float(fnr)
            if not np.isnan(fnr)
            else None,

        "ppv": float(ppv)
            if not np.isnan(ppv)
            else None,

        "npv": float(npv)
            if not np.isnan(npv)
            else None,

        "auc": float(auc)
            if not np.isnan(auc)
            else None,

        "selection_rate":
            float(selection_rate),

        "mean_predicted_risk":
            float(mean_predicted_risk),

        "n": int(len(y_true)),

        "positive_cases":
            int(np.sum(y_true == 1))
    }
```
